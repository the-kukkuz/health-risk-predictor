```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
heart_disease = fetch_ucirepo(id=45) 
  
# data (as pandas dataframes) 
X = heart_disease.data.features 
y = heart_disease.data.targets 
  
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.head())

print("\nTarget:")
print(y.head())

print("\nFeature columns:")
print(X.columns.tolist())

print("\nTarget columns:")
print(y.columns.tolist())

print("\nData types:")
print(X.dtypes)

print("\nMissing values:")
print(X.isnull().sum())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
X["sex"].value_counts()
```

```python
# Show unique values and their frequencies for every column

for column in X.columns:
    print(f"\n{'='*50}")
    print(f"{column}")
    print(f"{'='*50}")
    print(X[column].value_counts(dropna=False).sort_index())
```

```python
numerical_columns = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]

print(X[numerical_columns].describe())
```

```python
print(X[X.isnull().any(axis=1)])
```

```python
print("Duplicate rows:", X.duplicated().sum())
```

```python
import pandas as pd
categorical_columns = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]

for column in categorical_columns:
    print(f"\n{'='*60}")
    print(f"{column} vs target")
    print(f"{'='*60}")
    print(pd.crosstab(X[column], y["num"]))
```

```python
# Create a temporary binary target for EDA
eda_df = X.copy()

eda_df["disease"] = (y["num"] > 0).astype(int)

print(eda_df["disease"].value_counts())
```

```python
numerical_columns = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]

print(
    eda_df.groupby("disease")[numerical_columns].agg(
        ["count", "mean", "median", "std", "min", "max"]
    ).round(2)
)
```

```python
for column in numerical_columns:
    print(f"\n{'=' * 60}")
    print(f"{column} — No Disease vs Disease")
    print(f"{'=' * 60}")

    print(
        eda_df.groupby("disease")[column]
        .describe()
        .round(2)
    )
```

```python
print("\nDifference in group means:")
print(
    eda_df.groupby("disease")[numerical_columns]
    .mean()
    .round(2)
)
```
